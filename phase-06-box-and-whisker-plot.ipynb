# Phase 6 — Box and Whisker Plot

A **box plot** summarises the distribution of a dataset through its
quartiles: the box spans Q1–Q3, the line inside marks the median, and the
whiskers extend to the data range (excluding outliers).


```python
import matplotlib.pyplot as plt
```

## Basic Box Plot with Notch

The `notch` parameter adds a narrowed section around the median,
giving a rough sense of the median's confidence interval.


```python
x = [10, 20, 30, 40, 50, 60, 70]

plt.boxplot(x, notch=True)
plt.title("Box Plot Example")
plt.show()
```

## Horizontal Box Plot

Set `vert=False` to render the box plot horizontally.


```python
plt.boxplot(x, vert=False)
plt.title("Box Plot Example")
plt.show()
```

## Box Width

The `widths` parameter controls the width of the box.


```python
plt.boxplot(x, widths=0.5)
plt.title("Box Plot Example")
plt.show()
```

## Labels and Patch Artist

`labels` names each dataset. `patch_artist=True` fills the box with a
solid colour.


```python
plt.boxplot(x, labels=["Data Set 1"], patch_artist=True)
plt.title("Box Plot Example")
plt.show()
```

## Show Means

`showmeans=True` adds a marker at the mean value.


```python
plt.boxplot(x, labels=["Data Set 1"], showmeans=True)
plt.title("Box Plot Example")
plt.show()
```

## Multiple Datasets

Pass a list of lists to compare several distributions side by side.


```python
x1 = [10, 20, 30, 40, 50, 60, 70, 80, 120]
datasets = [x, x1]

plt.boxplot(datasets, labels=["Data Set 1", "Data Set 2"], showmeans=True)
plt.title("Multiple Box Plot Example")
plt.show()
```
