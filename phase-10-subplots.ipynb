# Phase 10 — Subplots

`plt.subplot(rows, cols, index)` creates a grid of smaller axes within a
single figure. This lets you display multiple plot types side by side
for comparison.


```python
import matplotlib.pyplot as plt
```

## 2×2 Grid of Different Plots

Each cell places a different chart type in one of the four panels:
line, pie, pie, and bar.


```python
# Top-left: line plot
x = [1, 2, 3, 4]
y1 = [1, 3, 2, 5]
plt.subplot(2, 2, 1)
plt.plot(x, y1, 'r')

# Top-right: pie chart
plt.subplot(2, 2, 2)
plt.pie([1], colors=['r'])

# Bottom-left: pie chart with data
x1 = [10, 20, 30, 40]
plt.subplot(2, 2, 3)
plt.pie(x1)

# Bottom-right: bar chart
x2 = ['a', 's', 'd', 'f']
y2 = [10, 20, 30, 40]
plt.subplot(2, 2, 4)
plt.bar(x2, y2)

plt.show()
```
