# Phase 5 — Stem Plot

A **stem plot** draws vertical lines from a baseline to each data point,
topped with a marker. It is useful for visualising discrete sequences
such as digital signals or event counts.


```python
import matplotlib.pyplot as plt
```

## Basic Stem Plot


```python
x = [1, 2, 3, 4, 5, 6]
y = [10, 12, 18, 20, 15, 8]

plt.stem(x, y)
plt.show()
```

## Line Style

Control the stem line format with `linefmt`.


```python
plt.stem(x, y, linefmt=':')
plt.show()
```

## Marker Format

`markerfmt` sets the marker shape and colour at the top of each stem.


```python
plt.stem(x, y, linefmt=':', markerfmt='r+')
plt.show()
```

```python
plt.stem(x, y, linefmt=':', markerfmt='ro')
plt.show()
```

## Base Format and Legend

`basefmt` styles the baseline. Add a label and call `plt.legend()` to
identify the series.


```python
plt.stem(x, y, linefmt=':', markerfmt='ro',
         basefmt='g-', label='stem plot')
plt.legend()
plt.show()
```
