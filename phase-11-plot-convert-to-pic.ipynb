# Convert plot graph to pic (jpg, png etc.)

```python
import matplotlib.pyplot as pyplot
```

```python
x = [1, 2, 3, 4, 5]
y = [2, 3, 5, 7, 11]

pyplot.plot(x, y)
pyplot.xlabel('X-axis Label')
pyplot.ylabel('Y-axis Label')
pyplot.title('Sample Line Plot')
pyplot.savefig('sample_plot3.png', dpi=300, facecolor='b', edgecolor='w', orientation='portrait')

pyplot.show()
```
