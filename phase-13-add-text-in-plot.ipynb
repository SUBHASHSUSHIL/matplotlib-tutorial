# Add text in plot

```python
import matplotlib.pyplot as plt
```

```python
x = [1, 2, 3, 4, 5]
y = [3, 1, 5, 2, 11]

plt.plot(x, y)
plt.xlabel('X-axis Label')
plt.ylabel('Y-axis Label')
plt.title('Sample Line Plot with Text', fontsize=18)

plt.text(3, 5, 'Important Point', fontsize=12, color='red', bbox=dict(facecolor='yellow', alpha=0.5))

plt.annotate('Peak', xy=(5, 11), xytext=(4, 10), arrowprops=dict(facecolor='black', shrink=0.05))

plt.legend(['Data Line'], loc='upper left')

plt.show()
```
