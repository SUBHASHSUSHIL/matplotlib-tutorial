# Box and Whisker Plot

```python
import matplotlib.pyplot as plt
```

```python
x = [10, 20, 30, 40, 50, 60, 70]

plt.boxplot(x, notch=True)
plt.title("Box Plot Example")

plt.show()
```

```python
plt.boxplot(x, vert=False)
plt.title("Box Plot Example")

plt.show()
```

```python
plt.boxplot(x, widths=0.5)
plt.title("Box Plot Example")

plt.show()
```

```python
plt.boxplot(x, labels=["Data Set 1"])
plt.title("Box Plot Example")

plt.show()
```

```python
plt.boxplot(x, labels=["Data Set 1"], patch_artist=True)
plt.title("Box Plot Example")

plt.show()
```

```python
plt.boxplot(x, labels=["Data Set 1"], showmeans=True)
plt.title("Box Plot Example")

plt.show() 
```

```python
x1 = [10, 20, 30, 40, 50, 60, 70, 80, 120]

y = [x, x1]

plt.boxplot(y, labels=["Data Set 1", "Data Set 2"], showmeans=True, )
plt.title("Multiple Box Plot Example")

plt.show() 
```
