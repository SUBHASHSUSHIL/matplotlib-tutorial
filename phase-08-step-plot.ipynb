# Phase 8 — Step Plot

A **step plot** draws data as a series of horizontal and vertical steps
rather than connecting points with diagonal lines. It is useful for
visualising data that changes at discrete intervals (e.g. digital
signals or price tiers).


```python
import matplotlib.pyplot as plt
```

## Basic Step Plot

The `where` parameter controls where the step occurs:
`'pre'` (step before the point), `'post'` (step after), or `'mid'`
(step at the midpoint).


```python
x = [1, 2, 3, 4, 5]
y1 = [1, 3, 2, 5, 4]

plt.step(x, y1, where='pre', label='Pre',
         linestyle='-', marker='o', ms=10, color='blue')
plt.grid()
plt.title('Step Plot Example')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.legend()
plt.show()
```
