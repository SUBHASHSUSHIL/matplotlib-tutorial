## Bar Plot

```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
x = ["python", "java", "c++", "ruby", "javascript"]
y = [92, 85, 78, 48, 88]

plt.bar(x, y)
plt.xlabel("Programming Languages")
plt.ylabel("Popularity Score")
plt.title("Popularity of Programming Languages")
plt.show()
```

```python
plt.bar(x, y)
plt.xlabel("Programming Languages", fontsize=14)
plt.ylabel("Popularity Score", fontsize=14)
plt.title("Popularity of Programming Languages", fontsize=16)
plt.show()
```

```python
plt.bar(x, y, width=0.2)
plt.xlabel("Programming Languages")
plt.ylabel("Popularity Score")
plt.title("Popularity of Programming Languages")
plt.show()
```

```python
c = ['r', 'b', 'green', 'orange', 'purple']
plt.bar(x, y, color=c)
plt.xlabel("Programming Languages")
plt.ylabel("Popularity Score")
plt.title("Popularity of Programming Languages")
plt.show()
```

```python
c = ['r', 'b', 'green', 'orange', 'purple']
plt.bar(x, y, color=c, align='edge')
plt.xlabel("Programming Languages")
plt.ylabel("Popularity Score")
plt.title("Popularity of Programming Languages")
plt.show()
```

```python
c = ['r', 'b', 'green', 'orange', 'purple']
plt.bar(x, y, color=c, edgecolor='black', label='Language Popularity')
plt.xlabel("Programming Languages")
plt.ylabel("Popularity Score")
plt.title("Popularity of Programming Languages")
plt.legend()
plt.show()
```

```python
x = ["python", "java", "c++", "ruby", "javascript"]
y = [92, 85, 78, 48, 88]
z = [80, 70, 75, 60, 90]
plt.xlabel("Programming Languages")
plt.ylabel("Popularity Score")
plt.title("Popularity of Programming Languages")
plt.bar(x, y, width=0.4, label='2023', align='edge')
plt.bar(x, z, width=-0.4, label='2022', align='edge')
plt.legend()
plt.show()
```

```python
x = ["python", "java", "c++", "ruby", "javascript"]
y = [92, 85, 78, 48, 88]
z = [80, 70, 75, 60, 90]

width = 0.4
p= np.arange(len(x))
p1 = [j+width for j in p]

plt.xlabel("Programming Languages", fontsize=14)
plt.ylabel("Popularity Score", fontsize=14)
plt.title("Popularity of Programming Languages", fontsize=16)

plt.bar(p, y, width=width, color='r', label='popularity')
plt.bar(p1, z, width=width, color='y', label='popularity1')

plt.xticks(p + width/2, x, rotation=45)
plt.legend()
plt.show()
```

```python
x = ["python", "java", "c++", "ruby", "javascript"]
y = [92, 85, 78, 48, 88]
z = [80, 70, 75, 60, 90]

width = 0.4
p= np.arange(len(x))
p1 = [j+width for j in p]

plt.xlabel("Programming Languages", fontsize=14)
plt.ylabel("Popularity Score", fontsize=14)
plt.title("Popularity of Programming Languages", fontsize=16)

plt.barh(p, y, height=width, color='r', label='popularity')
plt.barh(p1, z, height=width, color='y', label='popularity1')

plt.yticks(p + width/2, x, rotation=45)
plt.legend()
plt.show()
```
