## Bar Plot

```python
import matplotlib.pyplot as plt
```

```python
x = ["python", "java", "c++", "ruby", "javascript"]
y = [92, 85, 78, 48, 88]

plt.bar(x, y)
plt.xlabel("Programming Languages")
plt.ylabel("Popularity Score")
plt.title("Popularity of Programming Languages")
plt.show()
```
