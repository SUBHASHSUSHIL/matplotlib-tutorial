# Phase 13 — Text and Annotations

Matplotlib lets you add explanatory text and arrows directly onto a plot.
Use `plt.text()` for standalone labels and `plt.annotate()` for arrows
pointing to specific data points.


```python
import matplotlib.pyplot as plt
```

## Adding Text and an Annotation

- `plt.text(x, y, s)` places a text string at data coordinates `(x, y)`.
  The `bbox` parameter adds a background box.
- `plt.annotate()` draws an arrow from `xytext` to `xy`.
- `plt.legend()` identifies each plotted line.


```python
x = [1, 2, 3, 4, 5]
y = [3, 1, 5, 2, 11]

plt.plot(x, y)
plt.xlabel('X-axis Label')
plt.ylabel('Y-axis Label')
plt.title('Sample Line Plot with Text', fontsize=18)

plt.text(3, 5, 'Important Point', fontsize=12, color='red',
         bbox=dict(facecolor='yellow', alpha=0.5))
plt.annotate('Peak', xy=(5, 11), xytext=(4, 10),
             arrowprops=dict(facecolor='black', shrink=0.05))
plt.legend(['Data Line'], loc='upper left')
plt.show()
```
