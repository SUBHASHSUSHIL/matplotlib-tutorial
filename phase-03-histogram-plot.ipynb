# Phase 3 — Histogram Plot

A **histogram** groups continuous data into bins and shows the frequency
of values in each bin. It is useful for understanding the distribution
shape of a dataset.


```python
import matplotlib.pyplot as plt
import numpy as np
import random
```

## Generating Sample Data

Create 50 random integers in the range 10–80.


```python
data = np.random.randint(10, 80, 50)
print(data)
```

## Basic Histogram

Use `bins` to control the number of intervals and `edgecolor` to outline
each bar.


```python
data = [78, 18, 26, 52, 47, 11, 24, 69, 46, 38, 69, 16, 26, 41, 56, 36, 10, 11, 27, 66,
        66, 66, 52, 23, 72, 11, 45, 55, 39, 65, 18, 50, 72, 10, 43, 75, 59, 72, 29, 53,
        54, 19, 13, 35, 10, 67, 27, 63, 24, 47]

plt.hist(data, bins=8, color='purple', edgecolor='black')
plt.title('Histogram Plot')
plt.xlabel('Value')
plt.ylabel('Frequency')
plt.show()
```

## Auto Bins with Range

Set `bins='auto'` to let Matplotlib choose the bin count, and restrict
the visible range with `range`.


```python
data = [78, 18, 26, 52, 47, 11, 24, 69, 46, 38, 69, 16, 26, 41, 56, 36, 10, 11, 27, 66,
        66, 66, 52, 23, 72, 11, 45, 55, 39, 65, 18, 50, 72, 10, 43, 75, 59, 72, 29, 53,
        54, 19, 13, 35, 10, 67, 27, 63, 24, 47]

plt.hist(data, bins='auto', range=(0, 100), color='purple', edgecolor='black')
plt.title('Histogram Plot')
plt.xlabel('Value')
plt.ylabel('Frequency')
plt.show()
```

## Cumulative Histogram

Set `cumulative=-1` for a descending cumulative histogram (counts from
the top of the range downward).


```python
data = [78, 18, 26, 52, 47, 11, 24, 69, 46, 38, 69, 16, 26, 41, 56, 36, 10, 11, 27, 66,
        66, 66, 52, 23, 72, 11, 45, 55, 39, 65, 18, 50, 72, 10, 43, 75, 59, 72, 29, 53,
        54, 19, 13, 35, 10, 67, 27, 63, 24, 47]

plt.hist(data, bins='auto', range=(0, 100), color='purple',
         edgecolor='black', cumulative=-1)
plt.title('Histogram Plot')
plt.xlabel('Value')
plt.ylabel('Frequency')
plt.show()
```

## Custom Bin Edges and Hist Type

Provide an explicit list of bin edges and switch `histtype` to
`'stepfilled'` for a filled step appearance.


```python
data = [78, 18, 26, 52, 47, 11, 24, 69, 46, 38, 69, 16, 26, 41, 56, 36, 10, 11, 27, 66,
        66, 66, 52, 23, 72, 11, 45, 55, 39, 65, 18, 50, 72, 10, 43, 75, 59, 72, 29, 53,
        54, 19, 13, 35, 10, 67, 27, 63, 24, 47]
bins = [10, 20, 30, 40, 50, 60]

plt.hist(data, bins=bins, range=(0, 100), edgecolor='r',
         histtype='stepfilled')
plt.title('Histogram Plot')
plt.xlabel('Value')
plt.ylabel('Frequency')
plt.show()
```

## Horizontal Histogram

Set `orientation='horizontal'` to rotate the histogram 90 degrees.


```python
data = [78, 18, 26, 52, 47, 11, 24, 69, 46, 38, 69, 16, 26, 41, 56, 36, 10, 11, 27, 66,
        66, 66, 52, 23, 72, 11, 45, 55, 39, 65, 18, 50, 72, 10, 43, 75, 59, 72, 29, 53,
        54, 19, 13, 35, 10, 67, 27, 63, 24, 47]
bins = [10, 20, 30, 40, 50, 60]

plt.hist(data, bins=bins, range=(0, 100), edgecolor='r',
         orientation='horizontal', histtype='stepfilled')
plt.title('Histogram Plot')
plt.xlabel('Value')
plt.ylabel('Frequency')
plt.show()
```

## Relative Bar Width

Use `rwidth` to leave a gap between bars (value < 1).


```python
data = [78, 18, 26, 52, 47, 11, 24, 69, 46, 38, 69, 16, 26, 41, 56, 36, 10, 11, 27, 66,
        66, 66, 52, 23, 72, 11, 45, 55, 39, 65, 18, 50, 72, 10, 43, 75, 59, 72, 29, 53,
        54, 19, 13, 35, 10, 67, 27, 63, 24, 47]
bins = [10, 20, 30, 40, 50, 60]

plt.hist(data, bins=bins, range=(0, 100), edgecolor='r', rwidth=0.8)
plt.title('Histogram Plot')
plt.xlabel('Value')
plt.ylabel('Frequency')
plt.show()
```
