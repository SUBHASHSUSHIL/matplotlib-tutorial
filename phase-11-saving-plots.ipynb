# Phase 11 — Saving Plots as Image Files

Use `plt.savefig()` to export a figure to a file before calling
`plt.show()`. Supported formats include PNG, JPG, SVG, and PDF.


```python
import matplotlib.pyplot as plt
```

## Save a Line Plot to PNG

Key `savefig` parameters:
- `dpi` — resolution in dots per inch.
- `facecolor` — background colour of the figure.
- `edgecolor` — border colour.
- `orientation` — `'portrait'` or `'landscape'`.


```python
x = [1, 2, 3, 4, 5]
y = [2, 3, 5, 7, 11]

plt.plot(x, y)
plt.xlabel('X-axis Label')
plt.ylabel('Y-axis Label')
plt.title('Sample Line Plot')
plt.savefig('sample_plot.png', dpi=300)
plt.show()
```
