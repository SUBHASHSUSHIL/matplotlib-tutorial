# Phase 9 — Fill Between Plot

`plt.fill_between()` shades the area between two curves (or between a
curve and a constant). It is useful for highlighting ranges, confidence
intervals, or the difference between two datasets.


```python
import matplotlib.pyplot as plt
```

## Basic Fill Between


```python
x = [1, 2, 3, 4, 5]
y1 = [1, 3, 2, 5, 4]
y2 = [2, 4, 1, 3, 5]

plt.plot(x, y1, label='Line 1', color='blue')
plt.fill_between(x, y1, y2, color='lightblue', label='Fill Area')
plt.title('Fill Between Plot Example')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.legend()
plt.grid()
plt.show()
```
