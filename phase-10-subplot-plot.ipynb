# Subplot Plot

```python
import matplotlib.pyplot as plt
```

```python
x = [1, 2, 3, 4 ]
y1 = [1, 3, 2, 5]

plt.subplot(2, 2, 1)  # (rows, columns, panel number)
plt.plot(x, y1, 'r')  # red dashed line

plt.subplot(2, 2, 2)
plt.pie([1], colors="r")

x1 = [10, 20, 30, 40 ]
plt.subplot(2, 2, 3)
plt.pie(x1)

x2 = ["a", "s", "d", "f"]
y2 = [10, 20, 30, 40]
plt.subplot(2, 2, 4)
plt.bar(x2, y2)




plt.show()
```
