This notebook walks through setting up a Python environment for Matplotlib
and demonstrates the **bar plot**, one of the most common chart types for
comparing categorical data.

## Import

The standard convention is to import `pyplot` as `plt`.


````python
import matplotlib.pyplot as plt
import numpy as np
````

## Basic Bar Plot

A simple vertical bar chart comparing the popularity scores of five
programming languages.


````python
x = ["python", "java", "c++", "ruby", "javascript"]
y = [92, 85, 78, 48, 88]

plt.bar(x, y)
plt.xlabel("Programming Languages")
plt.ylabel("Popularity Score")
plt.title("Popularity of Programming Languages")
plt.show()
````

### Adjusting Font Sizes

Use `fontsize` to control label and title text size.


````python
plt.bar(x, y)
plt.xlabel("Programming Languages", fontsize=14)
plt.ylabel("Popularity Score", fontsize=14)
plt.title("Popularity of Programming Languages", fontsize=16)
plt.show()
````

### Bar Width

The `width` parameter controls the thickness of each bar (default 0.8).


````python
plt.bar(x, y, width=0.2)
plt.xlabel("Programming Languages")
plt.ylabel("Popularity Score")
plt.title("Popularity of Programming Languages")
plt.show()
````

### Custom Colors

Pass a list of color names to colour each bar individually.


````python
colors = ['r', 'b', 'green', 'orange', 'purple']
plt.bar(x, y, color=colors)
plt.xlabel("Programming Languages")
plt.ylabel("Popularity Score")
plt.title("Popularity of Programming Languages")
plt.show()
````

### Edge Alignment

Setting `align='edge'` aligns the bars to the right edge of the tick mark
instead of centring them.


````python
colors = ['r', 'b', 'green', 'orange', 'purple']
plt.bar(x, y, color=colors, align='edge')
plt.xlabel("Programming Languages")
plt.ylabel("Popularity Score")
plt.title("Popularity of Programming Languages")
plt.show()
````

### Edge Color and Legend

Add a border to bars with `edgecolor` and include a legend with `label`
plus `plt.legend()`.


````python
colors = ['r', 'b', 'green', 'orange', 'purple']
plt.bar(x, y, color=colors, edgecolor='black', label='Language Popularity')
plt.xlabel("Programming Languages")
plt.ylabel("Popularity Score")
plt.title("Popularity of Programming Languages")
plt.legend()
plt.show()
````

### Grouped (Side-by-Side) Bars

Plot two datasets side by side by offsetting the x-positions. Using
negative and positive `width` with `align='edge'` places bars to the
left and right of each tick.


````python
x = ["python", "java", "c++", "ruby", "javascript"]
y = [92, 85, 78, 48, 88]   # 2023 scores
z = [80, 70, 75, 60, 90]   # 2022 scores

plt.xlabel("Programming Languages")
plt.ylabel("Popularity Score")
plt.title("Popularity of Programming Languages")
plt.bar(x, y, width=0.4, label='2023', align='edge')
plt.bar(x, z, width=-0.4, label='2022', align='edge')
plt.legend()
plt.show()
````

### Grouped Bars with Proper Tick Alignment

A cleaner approach uses `np.arange` for x-positions and offsets each
group by half the bar width, then relabels the ticks with
`plt.xticks()`.


````python
x = ["python", "java", "c++", "ruby", "javascript"]
y = [92, 85, 78, 48, 88]
z = [80, 70, 75, 60, 90]
width = 0.4

x_pos = np.arange(len(x))
x_pos2 = [j + width for j in x_pos]

plt.xlabel("Programming Languages", fontsize=14)
plt.ylabel("Popularity Score", fontsize=14)
plt.title("Popularity of Programming Languages", fontsize=16)
plt.bar(x_pos, y, width=width, color='r', label='2023')
plt.bar(x_pos2, z, width=width, color='y', label='2022')
plt.xticks(x_pos + width / 2, x, rotation=45)
plt.legend()
plt.show()
````

### Horizontal Grouped Bars

Swap `bar` for `barh` (and `xticks` for `yticks`) to create a horizontal
version of the grouped chart.


````python
x = ["python", "java", "c++", "ruby", "javascript"]
y = [92, 85, 78, 48, 88]
z = [80, 70, 75, 60, 90]
width = 0.4

x_pos = np.arange(len(x))
x_pos2 = [j + width for j in x_pos]

plt.xlabel("Programming Languages", fontsize=14)
plt.ylabel("Popularity Score", fontsize=14)
plt.title("Popularity of Programming Languages", fontsize=16)
plt.barh(x_pos, y, height=width, color='r', label='2023')
plt.barh(x_pos2, z, height=width, color='y', label='2022')
plt.yticks(x_pos + width / 2, x, rotation=45)
plt.legend()
plt.show()
````
