# pie plot (Pie chart)

```python
import matplotlib.pyplot as plt
```

```python
x = [40, 23, 12, 30]
y = ["python", "c++", "java", "ruby"]

plt.pie(x, labels=y)

plt.show()
```

```python
ex = [0.0, 0.0, 0.2, 0.0]

plt.pie(x, labels=y, explode=ex)

plt.show()
```

```python
clr = ['r', 'b', 'g', 'y']

plt.pie(x, labels=y, explode=ex, colors=clr)

plt.show()
```

```python
plt.pie(x, labels=y, explode=ex, colors=clr, autopct='%1.1f%%')

plt.show()
```

```python
plt.pie(x, labels=y, explode=ex, colors=clr, autopct='%1.1f%%', shadow=True)

plt.show()
```

```python
plt.pie(x, labels=y, explode=ex, colors=clr, autopct='%1.1f%%', shadow=True, radius=1.2)

plt.show()
```

```python
plt.pie(x, labels=y, explode=ex, colors=clr, autopct='%1.1f%%', shadow=True, radius=1.2, labeldistance=1.1)

plt.show()
```

```python
plt.pie(x, labels=y, explode=ex, colors=clr, autopct='%1.1f%%', shadow=True, radius=1.2, labeldistance=1.1, startangle=250)

plt.show()
```

```python
plt.pie(x, labels=y, explode=ex, colors=clr, autopct='%1.1f%%', shadow=True, radius=1.2, labeldistance=1.1, startangle=250
        , textprops={'fontsize': 15})

plt.show()
```

```python
plt.pie(x, labels=y, explode=ex, colors=clr, autopct='%1.1f%%', shadow=True, radius=1.2, labeldistance=1.1, startangle=250
        , textprops={'fontsize': 15}, counterclock=False)

plt.show()
```

```python
plt.pie(x, labels=y, explode=ex, colors=clr, autopct='%1.1f%%', shadow=True, radius=1.2, labeldistance=1.1, startangle=250
        , textprops={'fontsize': 15}, counterclock=False, wedgeprops={"linewidth": 5})

plt.show()
```

```python
plt.pie(x, labels=y, explode=ex, colors=clr, autopct='%1.1f%%', shadow=True, radius=1.2, labeldistance=1.1, startangle=250
        , textprops={'fontsize': 15}, counterclock=False, wedgeprops={"linewidth": 5, "width": 1})

plt.show()
```

```python
plt.pie(x, labels=y, explode=ex, colors=clr, autopct='%1.1f%%', shadow=True, radius=1.2, labeldistance=1.1, startangle=250
        , textprops={'fontsize': 15}, counterclock=False, wedgeprops={"linewidth": 5, "width": 1}, center=(1, 1))

plt.show()
```

```python
plt.pie(x, labels=y, explode=ex, colors=clr, autopct='%1.1f%%', shadow=True, radius=1.2, labeldistance=1.1, startangle=250
        , textprops={'fontsize': 15}, counterclock=False, wedgeprops={"linewidth": 5, "edgecolor": "m"})

plt.show()
```

```python
plt.pie(x, labels=y, explode=ex, colors=clr, autopct='%1.1f%%', shadow=True, radius=1.2, labeldistance=1.1, startangle=250
        , textprops={'fontsize': 15}, counterclock=False, wedgeprops={"linewidth": 5, "edgecolor": "m"})

plt.title("My Pie Chart")
plt.axis('equal')
plt.legend(loc=2)

plt.show()
```
