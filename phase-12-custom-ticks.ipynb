# Phase 12 — Customising Tick Labels

`plt.xticks()` and `plt.yticks()` let you replace numeric tick positions
with custom text labels, giving you full control over axis annotations.


```python
import matplotlib.pyplot as plt
```

## Replacing Numeric Ticks with Text Labels


```python
x = [1, 2, 3, 4, 5]
y = [2, 3, 5, 7, 11]

plt.plot(x, y)
plt.xticks([1, 2, 3, 4, 5], ['One', 'Two', 'Three', 'Four', 'Five'])
plt.yticks([2, 3, 5, 7, 11], ['Two', 'Three', 'Five', 'Seven', 'Eleven'])
plt.xlabel('X-axis Label')
plt.ylabel('Y-axis Label')
plt.title('Line Plot with Custom Ticks')
plt.show()
```
