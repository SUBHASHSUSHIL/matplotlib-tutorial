# Step Plot

```python
import matplotlib.pyplot as plt
```

```python
x = [1, 2, 3, 4, 5]
y1 = [1, 3, 2, 5, 4]

plt.step(x, y1, where='pre', label='Pre', linestyle='-', marker='o', ms=10, color='blue')
plt.grid()
plt.title('Step Plot Example')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.legend()

plt.show()
```
