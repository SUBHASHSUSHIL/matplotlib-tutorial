# Phase 4 — Pie Plot

A **pie chart** represents proportions of a whole. Each wedge's size is
proportional to the value it represents.


```python
import matplotlib.pyplot as plt
```

## Basic Pie Chart


```python
x = [40, 23, 12, 30]
y = ["python", "c++", "java", "ruby"]

plt.pie(x, labels=y)
plt.show()
```

## Explode (Highlight a Slice)

The `explode` parameter offsets a slice from the centre. Values are
fractions of the radius.


```python
ex = [0.0, 0.0, 0.2, 0.0]
plt.pie(x, labels=y, explode=ex)
plt.show()
```

## Custom Colours


```python
clr = ['r', 'b', 'g', 'y']
plt.pie(x, labels=y, explode=ex, colors=clr)
plt.show()
```

## Percentage Labels

Use `autopct` to display the percentage value on each wedge.


```python
plt.pie(x, labels=y, explode=ex, colors=clr, autopct='%1.1f%%')
plt.show()
```

## Shadow Effect


```python
plt.pie(x, labels=y, explode=ex, colors=clr,
        autopct='%1.1f%%', shadow=True)
plt.show()
```

## Radius and Label Distance

`radius` scales the entire pie; `labeldistance` moves labels closer or
farther from the centre.


```python
plt.pie(x, labels=y, explode=ex, colors=clr, autopct='%1.1f%%',
        shadow=True, radius=1.2, labeldistance=1.1)
plt.show()
```

## Start Angle and Text Properties

`startangle` rotates the beginning of the first wedge. `textprops`
controls label font properties.


```python
plt.pie(x, labels=y, explode=ex, colors=clr, autopct='%1.1f%%',
        shadow=True, radius=1.2, labeldistance=1.1,
        startangle=250, textprops={'fontsize': 15})
plt.show()
```

## Counterclock and Wedge Properties

`counterclock=False` draws wedges clockwise. `wedgeprops` controls the
wedge border line width and edge colour.


```python
plt.pie(x, labels=y, explode=ex, colors=clr, autopct='%1.1f%%',
        shadow=True, radius=1.2, labeldistance=1.1,
        startangle=250, textprops={'fontsize': 15},
        counterclock=False, wedgeprops={'linewidth': 5, 'edgecolor': 'm'})
plt.show()
```

## Title, Equal Aspect, and Legend

Add a title, ensure the pie is drawn as a circle with `plt.axis('equal')`,
and show a legend.


```python
plt.pie(x, labels=y, explode=ex, colors=clr, autopct='%1.1f%%',
        shadow=True, radius=1.2, labeldistance=1.1,
        startangle=250, textprops={'fontsize': 15},
        counterclock=False, wedgeprops={'linewidth': 5, 'edgecolor': 'm'})
plt.title("My Pie Chart")
plt.axis('equal')
plt.legend(loc=2)
plt.show()
```

## Donut Chart

Overlay a white circle on top of a pie chart to create a donut effect.


```python
plt.pie(x, labels=y, radius=1.5)
plt.pie([1], colors=['w'])
plt.show()
```
