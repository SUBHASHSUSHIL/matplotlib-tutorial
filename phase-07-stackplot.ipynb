# Phase 7 — Stack Plot (Area Chart)

A **stack plot** stacks multiple datasets on top of one another, showing
both individual contributions and the cumulative total. It is great for
visualising how parts make up a whole over time.


```python
import matplotlib.pyplot as plt
```

## Basic Stack Plot


```python
x = [1, 2, 3, 4, 5]
y1 = [1, 3, 2, 5, 4]
y2 = [2, 4, 1, 3, 5]
y3 = [3, 2, 4, 1, 6]

plt.stackplot(x, y1, y2, y3,
              labels=['Y1', 'Y2', 'Y3'],
              colors=['#72f408', '#66b3ff', '#99ff99'])
plt.legend(loc='upper left')
plt.title('Stack Plot Example')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.show()
```
