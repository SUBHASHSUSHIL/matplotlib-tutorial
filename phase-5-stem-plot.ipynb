# Stem plot

```python
import matplotlib.pyplot as plt
```

```python
x = [1, 2, 3, 4, 5, 6]
y = [10, 12, 18, 20, 15, 8]

plt.stem(x, y)
plt.show()
```

```python
plt.stem(x, y, linefmt=':')
plt.show()
```

```python
plt.stem(x, y, linefmt=':', markerfmt='r+')
plt.show()
```

```python
plt.stem(x, y, linefmt=':', markerfmt='ro')
plt.show()
```

```python
plt.stem(x, y, linefmt=':', markerfmt='r--', basefmt='g-', label='stem plot')
plt.legend()
plt.show()
```
